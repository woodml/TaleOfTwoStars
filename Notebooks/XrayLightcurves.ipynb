## Long-term lightcurve

In this notebook, we collect the X-ray fluxes from the literature and the Chandra and XMM data reductions done elsewhere and make them into a plot.

Data is manually input here and hard-coded in the notebook, not read feom the outputs of the previous notebooks.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
from astropy.time import Time
t_PSPC = Time('1992-09-22', format='iso')
t_RASS = Time('1990-08-27', format='iso')
t_XMM_2002 = Time(52147, format='mjd')
t_XMM_2012 = Time(55973.5, format='mjd')
t_XMM_slew = Time('2013-02-17', format='iso')
t_Chandra = Time(52970.4112, format='mjd')
```

```python
## SU Aurigae
# Use the following parameters for webPIMMS
# n_H = 0.44e22 (mean of the numbers in table 4)
# log kT = 7
# 0.3-10.0 keV
xmm_slew2_flux = 1.473e-12  # absorbed flux
xmm_slew2_flux = 3.55e-12 # unabsorbed flux
# For log t [MK] = 7.3
# 2.3e-12 and 3.9e-12


pspc2flux = 1.06e-11  # absorbed flux
pspc2flux = 2.5e-11 # unabsorbed flux
# for log T = 7.3
# 2.0 e-11 and 3.4 e-11

## AB Aurigae
# And similar for AB Aur (only listing the conversion for unabsorbed flux because
# it's almost the same anyway)
# using N_H = 0.23e20 from Table 9, 2012 data.
# Temperature   log T = 7 (the dominating component in both 2002 and 2012)
xmm_slew2_flux_ab = 1.2e-12


pspc2flux_ab = 7.3e-12
# Using the cooler temperature this would be 6e-12
```

```python
d2 = 4 * np.pi * (156.0 * 3.086e18)**2  # distance in cm
```

```python
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(4,4,), sharex=True)

ax = axes[0]

axt = ax.twiny()

def MJD2year(mjd):
    return (mjd - 51544) / 365.25 + 2000

def convert_ax_mjd_to_year(ax):
    x1, x2 = ax.get_xlim()
    axt.set_xlim(MJD2year(x1), MJD2year(x2))
    axt.figure.canvas.draw()
    return ax

ax.callbacks.connect('xlim_changed', convert_ax_mjd_to_year)

yfactor = 1e-30
# 2RXS
ax.errorbar(t_RASS.mjd,
              0.13 * pspc2flux * d2 * yfactor,
              yerr=0.02 * pspc2flux  * d2* yfactor,
                fmt='s', label='RASS', color='k')
# 2nd PSPC catalog
ax.errorbar(t_PSPC.mjd,
             2.13e-1 * pspc2flux * d2 * yfactor,
             yerr=1.3e-2 * pspc2flux * d2 * yfactor,
             fmt='+', label='2nd PSPC\ncatalog', color='k')
# Table 4: quiescent
ax.errorbar([t_XMM_2002.mjd, t_XMM_2012.mjd], 
             np.array([3.4e-12, 3.7e-12]) * d2 * yfactor, 
             yerr=np.array([[0.1e-12, 0.4e-12], [0.4e-12, 0.4e-12]]) * d2 * yfactor,
               fmt='3', label='XMM', color='k')
# Table 7
ax.errorbar(t_Chandra.mjd,
            2.6e-12 * d2 * yfactor, 
            yerr=np.array([[0.2e-12], [0.4e-12]]) * d2 * yfactor, fmt='>', label='Chandra', color='k')
# Section 2.1.1
ax.errorbar(t_XMM_slew.mjd,  
             1 * xmm_slew2_flux * d2 * yfactor, 
             yerr=np.array([[0.6], [1.4]]) * d2 * xmm_slew2_flux * yfactor, 
             fmt='v', label='XMM slew', color='k')
ax.set_ylabel(r'$L_\mathrm{X}$ [$10^{30}$ erg/s]')

ax2 = axes[1]
#yfactor = 1e-29
# 2RXS
ax2.errorbar(t_PSPC.mjd,
             7.71e-3 * pspc2flux_ab * d2 * yfactor,
              yerr=2.65e-3 * pspc2flux_ab * d2 * yfactor,
              fmt='+', label='2nd PSPC catalog', color='k')
# table 9
ax2.errorbar([t_XMM_2002.mjd, t_XMM_2012.mjd], 
              np.array([2.5e-13, 2.1e-13]) * d2 * yfactor,
              yerr=np.array([[0.6e-13,0.6e-13], [.4e-13, .4e-13]]) * d2 * yfactor,
              fmt='3', label='XMM', color='k')
# section 4.2
ax2.errorbar(t_Chandra.mjd,
              1.8e-13 * d2 * yfactor,
              yerr= np.array([[0.5e-13], [0.4e-13]])* d2 * yfactor,
                fmt='>', label='Chandra', color='k')
# Section 2.1.1
ax2.errorbar(t_XMM_slew.mjd,
              0.24 * xmm_slew2_flux_ab * d2 * yfactor,
              yerr=np.array([[0.24], [0]]) * d2 * xmm_slew2_flux_ab * yfactor,
              fmt='v', label='XMM slew', color='k')
ax2.set_ylabel(r'$L_\mathrm{X}$ [$10^{30}$ erg/s]')

ax2.set_xlabel('MJD')
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
ax.text(0.1, 0.9, 'SU Aur', transform=ax.transAxes, ha='center')
ax2.text(0.1, 0.9, 'AB Aur', transform=ax2.transAxes, ha='center')
ax2.set_ylim(0, None)

fig.savefig('../Figures/long-term-lightcurve.pdf', bbox_inches='tight')
```
