```python
ui.set_source('2002_emos1', ui.xsphabs.abs1*(ui.xsvapec.v1))
ui.set_source('2002_emos2', abs1*(v1))

# - - - - - - - - - - - - - - - - - - - - - - - - - - - - - - - - - 
# Set initial guesses
abs1.nH = 0.1
v1.kT   = 1.0
v1.norm = 0.001

v1.O  = 1.0    # mid-FIP
v1.Ne = 1.0    # high-FIP
v1.Fe = 1.0    # low-FIP

# freeze/thaw
abs1.nH.min = 1e-4
abs1.nH.frozen = 0
v1.norm.frozen = 0
v1.kT.frozen = 0

v1.O.frozen  = 0
v1.Fe.frozen = 0
v1.Ne.frozen = 0

# Tie abundances
v1.Al = v1.Fe
v1.Ca = v1.Fe
v1.Ni = v1.Fe
v1.Mg = v1.Fe
v1.Si = v1.Fe

v1.C = v1.O
v1.N = v1.O
v1.S = v1.O

# - - - - - - - - - - - - - - - - - - - - - - - - - - - - - - - - - 
ui.set_method('levmar')
ui.fit('2002_emos1', '2002_emos2')
```

```python
plt.figure(figsize=(10,3.5))
ui.plot_fit('2002_emos1')
ui.plot_fit('2002_emos2', overplot=True)

g = plt.gca()
g.lines[1].set_label('EMOS 1')
g.lines[1].set_markersize(10)
g.lines[1].set_markeredgecolor('k')
g.lines[5].set_label('EMOS 2')
g.lines[5].set_markersize(10)
g.lines[5].set_markeredgecolor('k')

plt.xlabel(r'Wavelength ($\AA$)')
plt.ylabel(r'Counts/s/$\AA$')
plt.legend()
plt.title('')
plt.show()
```

```python
ui.covar('2002_emos1', '2002_emos2')
cov = ui.get_covar_results()
```

```python
# Write out
# Create table
output_2002 = Table(names=('Parameter', 'Value','e_lower','e_upper'), dtype=('str','float','float','float'))

# Get error dictionaries
cov_min = dict(zip(list(cov.parnames), list(cov.parmins)))
cov_max = dict(zip(list(cov.parnames), list(cov.parmaxes)))


for p in abs1.pars:
    parname = str(abs1.name[-4:]) + '.' + str(p.name)
    if parname in cov_min.keys():
        e_lower = cov_min[parname]
        e_upper = cov_max[parname]
        output_2002.add_row([parname, float(p.val), e_lower, e_upper])
    else:
        output_2002.add_row([parname, float(p.val), float('inf'), float('inf')])

i = 0
for v in (v1):
    for p in v.pars:
        parname = str(v.name[-2:]) + '.' + str(p.name)
        if p.name in ['He','C','N','O','Ne','Mg','Al','Si','S','Ar','Ca','Fe','Ni'] and i > 0:
            continue
        if parname in cov_min.keys():
            e_lower = cov_min[parname]
            e_upper = cov_max[parname]
            output_2002.add_row([parname, float(p.val), e_lower, e_upper])
        else:
            output_2002.add_row([parname, float(p.val), float('inf'), float('inf')])
    i += 1

output_2002['Value'].format= '.6f'
output_2002['e_lower'].format= '.5f'
output_2002['e_upper'].format= '.5f'

# norm = 1e-14* EM /(4*pi*d^2)
# EM = 4*pi*d^2 * norm / 1 e-14
d = 1000/6.3700 * u.pc  # parallax is 6.37 mas (Gaia DR3)
d = d.to(u.cm)

for row in output_2002:
    if 'norm' in row['Parameter']:
        v = row['Parameter'][:2] + '.EM'
        # norm = EM * 10**-14/(4*np.pi*d**2)
        em = row['Value']/(10**-14)*(4*np.pi*d**2)
        e_plus = row['e_upper']/(10**-14)*(4*np.pi*d**2)
        e_minus = row['e_lower']/(10**-14)*(4*np.pi*d**2)
        output_2002.add_row([v, em/1e53, e_minus/1e53, e_plus/1e53])
    elif 'kT' in row['Parameter']:
        v = row['Parameter'][:2] + '.T'
        # kT = k_B * T
        T = (row['Value'] / c.k_B.to('keV/K')).value
        t_minus = (row['e_lower'] / c.k_B.to('keV/K')).value
        t_plus = (row['e_upper'] / c.k_B.to('keV/K')).value
        output_2002.add_row([v, T/1e6, t_minus/1e6, t_plus/1e6])

output_2002.pprint_all()

output_2002.write('/Users/woodml/Documents/X-Ray/TaleofTwoStars/AB Aur/2002_fit_results.txt', format='latex', overwrite=True)
```

```python
# Write out
# Create table
output_2012 = Table(names=('Parameter', 'Value','e_lower','e_upper'), dtype=('str','float','float','float'))

# Get error dictionaries
cov_min = dict(zip(list(cov.parnames), list(cov.parmins)))
cov_max = dict(zip(list(cov.parnames), list(cov.parmaxes)))


for p in abs1.pars:
    parname = str(abs1.name[-4:]) + '.' + str(p.name)
    if parname in cov_min.keys():
        e_lower = cov_min[parname]
        e_upper = cov_max[parname]
        output_2012.add_row([parname, float(p.val), e_lower, e_upper])
    else:
        output_2012.add_row([parname, float(p.val), float('inf'), float('inf')])

i = 0
for v in (v1,v2):
    for p in v.pars:
        parname = str(v.name[-2:]) + '.' + str(p.name)
        if p.name in ['He','C','N','O','Ne','Mg','Al','Si','S','Ar','Ca','Fe','Ni'] and i > 0:
            continue
        if parname in cov_min.keys():
            e_lower = cov_min[parname]
            e_upper = cov_max[parname]
            output_2012.add_row([parname, float(p.val), e_lower, e_upper])
        else:
            output_2012.add_row([parname, float(p.val), float('inf'), float('inf')])
    i += 1

output_2012['Value'].format= '.6f'
output_2012['e_lower'].format= '.5f'
output_2012['e_upper'].format= '.5f'

# norm = 1e-14* EM /(4*pi*d^2)
# EM = 4*pi*d^2 * norm / 1 e-14
d = 1000/6.3700 * u.pc  # parallax is 6.37 mas (Gaia DR3)
d = d.to(u.cm)

for row in output_2012:
    if 'norm' in row['Parameter']:
        v = row['Parameter'][:2] + '.EM'
        # norm = EM * 10**-14/(4*np.pi*d**2)
        em = row['Value']/(10**-14)*(4*np.pi*d**2)
        e_plus = row['e_upper']/(10**-14)*(4*np.pi*d**2)
        e_minus = row['e_lower']/(10**-14)*(4*np.pi*d**2)
        output_2012.add_row([v, em/1e53, e_minus/1e53, e_plus/1e53])
    elif 'kT' in row['Parameter']:
        v = row['Parameter'][:2] + '.T'
        # kT = k_B * T
        T = (row['Value'] / c.k_B.to('keV/K')).value
        t_minus = (row['e_lower'] / c.k_B.to('keV/K')).value
        t_plus = (row['e_upper'] / c.k_B.to('keV/K')).value
        output_2012.add_row([v, T/1e6, t_minus/1e6, t_plus/1e6])

output_2012.pprint_all()

output_2012.write('/Users/woodml/Documents/X-Ray/TaleofTwoStars/AB Aur/2012_fit_results.txt', format='latex', overwrite=True)
```
