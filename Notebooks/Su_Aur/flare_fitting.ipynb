```python
# Write out
# Create table
output_2002 = Table(names=('Parameter', 'Value','e_lower','e_upper'), dtype=('str','float','float','float'))

# Get error dictionaries
cov_min = dict(zip(list(cov.parnames), list(cov.parmins)))
cov_max = dict(zip(list(cov.parnames), list(cov.parmaxes)))

for a in (abs1): 
    for p in a.pars:
        parname = str(a.name[-4:]) + '.' + str(p.name)
        if parname in cov_min.keys():
            e_lower = cov_min[parname]
            e_upper = cov_max[parname]
            output_2002.add_row([parname, float(p.val), e_lower, e_upper])
        else:
            output_2002.add_row([parname, float(p.val), float('inf'), float('inf')])

i = 0
for v in (v1, v2, v3):
    for p in v.pars:
        parname = str(v.name[-2:]) + '.' + str(p.name)
        if p.name in ['He','C','N','O','Ne','Mg','Al','Si','S','Ar','Ca','Fe','Ni'] and i > 0:
            continue
        if parname in cov_min.keys():
            e_lower = cov_min[parname]
            e_upper = cov_max[parname]
            output_2002.add_row([parname, float(p.val), e_lower, e_upper])
        else:
            output_2002.add_row([parname, float(p.val), float('inf'), float('inf')])
    i += 1

output_2002['Value'].format= '.3f'
output_2002['e_lower'].format= '.3f'
output_2002['e_upper'].format= '.3f'

# norm = 1e-14* EM /(4*pi*d^2)
# EM = 4*pi*d^2 * norm / 1 e-14
d = 1000/6.3700 * u.pc  # parallax is 6.37 mas (Gaia DR3)
d = d.to(u.cm)

for row in output_2002:
    if 'norm' in row['Parameter']:
        v = row['Parameter'][:2] + '.EM'
        # norm = EM * 10**-14/(4*np.pi*d**2)
        em = row['Value']/(10**-14)*(4*np.pi*d**2)
        e_plus = row['e_upper']/(10**-14)*(4*np.pi*d**2)
        e_minus = row['e_lower']/(10**-14)*(4*np.pi*d**2)
        output_2002.add_row([v, em/1e53, e_minus/1e53, e_plus/1e53])
    elif 'kT' in row['Parameter']:
        v = row['Parameter'][:2] + '.T'
        # kT = k_B * T
        T = (row['Value'] / c.k_B.to('keV/K')).value
        t_minus = (row['e_lower'] / c.k_B.to('keV/K')).value
        t_plus = (row['e_upper'] / c.k_B.to('keV/K')).value
        output_2002.add_row([v, T/1e6, t_minus/1e6, t_plus/1e6])

output_2002.pprint_all()

output_2002.write('/Users/woodml/Documents/X-Ray/TaleofTwoStars/SU_Aur/2002_flare_fit_results.txt', format='latex', overwrite=True)
```
